```python
# Challenge 1 — Multiplication Table
# Take a number from the user and print its multiplication table from 1 to 10 using a for loop.


num = int(input("Enter a number : "))

for i in range(1,11):
        print(num," * ",i," = ",num*i)
```

```python
# Challenge 2 — Sum of Even Numbers
# Take an integer n and calculate the sum of all even numbers from 1 through n using a loop. Handle n below 1.

num = int(input("Enter the number : "))

if num < 1:
    print("Please enter a number 1 or above.")
else:
    total_sum = 0
    for i in range(1, num + 1):
        if i % 2 == 0:
            total_sum += i
    print("Sum of all even numbers from 1 to", num, "is", total_sum)
            
```

```python
# Challenge 3 — Reverse a Number
# Take a positive integer and reverse its digits using a while loop.

num = int(input("Enter a number : "))

if num < 1:
    print("Please enter a number 1 or above.")
else:
    reversed_num = 0
    temp = num

    while temp > 0:
        digit = temp % 10
        reversed_num = (reversed_num * 10) + digit
        temp //= 10

    print(f"Reversed number of {num} is {reversed_num}")
```
