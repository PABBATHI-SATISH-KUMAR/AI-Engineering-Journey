```python
#Challenge 1 — Employee Profile

#Create variables for an employee's name, age, company, department, and monthly salary. Print them using f-strings.
```

```python
employee_name = "Satish"
employee_age = 23
employee_company = "ADP"
employee_Department = "DIS"
employee_salary = 20000

print("Employee_Name :- " f"{employee_name}")
print("Employee_age :- " f"{employee_age}")
print("Employee_company :- " f"{employee_company}")
print("Employee_Department :- " f"{employee_Department}")
print("Employee_salary :- " f"{employee_salary}")


print(f"Employee Name: {employee_name}")
print(f"Employee Age: {employee_age}")
print(f"Company: {employee_company}")
print(f"Department: {employee_Department}")
print(f"Salary: ₹{employee_salary}")
```

```python
# Challenge 2 — Salary Increment

# An employee earns ₹40,000 per month and receives a 15% salary increment. Calculate the updated salary and annual salary using variables.
```

```python
# emp_salary = 40000
# updated_Salary = emp_salary + (emp_salary * 0.15)
# annual_Salary = (emp_salary + updated_Salary) * 12

# print("Actual salary : ",emp_salary)
# print("updated_Salary : " ,updated_Salary)
# print("annual_Salary : "  ,annual_Salary)


emp_salary = 40000
increment_percentage = 15

updated_salary = emp_salary * (1 + increment_percentage / 100)
annual_salary = updated_salary * 12

print(f"Original Salary: ₹{emp_salary}")
print(f"Updated Salary: ₹{updated_salary}")
print(f"Annual Salary: ₹{annual_salary}")
```

```python
# Challenge 3 — Reference Behavior

# Predict the output before executing:
```

```python
a = [10, 20, 30]
b = a
c = a.copy()

a.append(40)
b.append(50)

print(a)
print(b)
print(c)
#because c is a shallow copy , changes on a and b won't affect c
```

```python
# Challenge 4 — Swapping

# Write a program to swap two variables without using a third variable.
```

```python
x , y = 10, 20 
print("Before swapping : ", "x : ",x, "and y : ",y)
x,y = y,x
print("After swapping : ", "x : ",x, "and y : ",y)
```
