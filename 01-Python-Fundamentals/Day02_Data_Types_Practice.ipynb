```python
# Challenge 1 — Identify data types
# Create variables for your name, age, salary, employment status, skills, and employee details. Print the type of each variable using type().
```

```python
name = "Satish"
age = 23
salary = 45000.50
emp_status = "Active"
skills = ["Python","Java"]
emp_details  = {
    "name" : "Satish",
    "age" : 23,
    "salary" : 45000.50,
    "emp_status" : "Active",
    "skills" : ["Python","Java"]
}

print(type(name))
print(type(age))
print(type(salary))
print(type(emp_status))
print(type(skills))
print(type(emp_details))
```

```python
# Challenge 2 — Data type conversion
# Given:
# age = "23"salary = "45000.75"experience = "2"



# Convert them into appropriate numeric types, calculate the employee's annual salary, and print the converted types.
```

```python
age = "23"
salary = "45000.75"
experience = "2"

age = int(age)
salary = float(salary)
experience = int(experience)

annual_Salary = salary * 12

print(age)
print(salary)
print(experience)
print(annual_Salary)
```
