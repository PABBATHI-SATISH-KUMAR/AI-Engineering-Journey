```python
# Challenge 1 — Salary calculation
# An employee earns ₹35,000 monthly, receives a 12% increment, and has a monthly deduction of ₹2,500.

# Calculate the updated monthly salary after deduction and the annual salary.
```

```python
# Initial parameters
initial_monthly_salary = 35000
increment_percentage = 12
monthly_deduction = 2500

# Calculations
increment_amount = initial_monthly_salary * (increment_percentage / 100)
gross_monthly_salary = initial_monthly_salary + increment_amount
updated_monthly_salary = gross_monthly_salary - monthly_deduction

annual_net_salary = updated_monthly_salary * 12
annual_gross_salary = gross_monthly_salary * 12

# Output results
print(f"Updated Monthly Salary (after deduction): ₹{updated_monthly_salary:,.2f}")
print(f"Annual Salary (Net): ₹{annual_net_salary:,.2f}")
print(f"Annual Salary (Gross): ₹{annual_gross_salary:,.2f}")
```

```python
# Challenge 3 — Employee eligibility
# An employee is eligible for a technical role when all conditions are satisfied:
# - Experience is at least 2 years.
# - The employee knows Python.
# - The employee is currently available.
# Use comparison, membership, and logical operators to check eligibility.
```

```python
# Candidate details
experience_years = 3
skills = ["Python", "SQL", "Docker"]
is_available = True

# Eligibility check
# 1. Comparison operator: >=
# 2. Membership operator: in
# 3. Logical operator: and
is_eligible = (experience_years >= 2) and ("Python" in skills) and is_available

# Output result
if is_eligible:
    print("Eligible for the technical role.")
else:
    print("Not eligible for the technical role.")
```
