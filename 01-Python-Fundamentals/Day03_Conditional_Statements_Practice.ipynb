```python
# Challenge 1 — Positive, Negative, or Zero

# Take an integer from the user and print whether it is positive, negative, or zero using if-elif-else.
```

```python
num = int(input("Enter a number : "))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# Challenge 2 — Student Grade Calculator

# Take marks between 0 and 100 and calculate the grade:
# Marks	Grade
# 90–100	A+
# 80–89	A
# 70–79	B
# 60–69	C
# 40–59	D
# Below 40	F

# Also handle invalid marks outside 0–100.
```

```python
marks = int(input("Enter the marks of the student : "))

if not (0 <= marks >100):
    print("Invalid marks ")
elif marks >= 90 or marks <= 100:
    print("A+")
elif marks >= 80 or marks <= 89:
    print("A")
elif marks >= 70 or marks <= 79:
    print("B")
elif marks >= 60 or marks <= 69:
    print("C")
elif marks >=40 or marks <=59:
    print("D")
else:
    print("F")
```

```python
# Challenge 3 — Employee Bonus Calculator

# Take an employee's salary and performance rating (1–5).
# - Rating 4.5 or above: 20% bonus.
# - Rating 4.0–4.49: 15% bonus.
# - Rating 3.0–3.99: 10% bonus.
# - Below 3.0: No bonus.
# Print the bonus amount and updated salary. Validate the salary and rating.
```

```python
emp_salary = int(input("Enter the salary of the employee : "))
rating = float(input("Enter the rating of the employee : "))

if not (0 <= rating > 5):
    print("Enter valid ratings [i.e., in between 1-5 ]")

elif rating >= 4.5:
    bonus = 20/100
    bonus_amount = emp_salary * bonus
    updated_salary = emp_salary + bonus_amount
    print(bonus_amount)
    print(updated_salary)
elif rating >= 4.0 or rating <= 4.49:
    bonus = 15/100
    bonus_amount = emp_salary * bonus
    updated_salary = emp_salary + bonus_amount
    print(bonus_amount)
    print(updated_salary)
elif rating >= 3.0 or rating <= 3.99:
    bonus = 10/100
    bonus_amount = emp_salary * bonus
    updated_salary = emp_salary + bonus_amount
    print(bonus_amount)
    print(updated_salary)
else:
    print("no bonus")
    
```
